# A simple random number generator function

```python
class random0:
    """This implements a LCG pRNG.
    
    seed : the seed of the pRNG sequence
    """
    seed: int = 12345
    a,c,m = (8121,28411,134456)
    def fire(self):
        """Generate a random number"""
        self.seed = (self.a*self.seed+self.c)%self.m
        return self.seed/self.m
    
    
r = random0()
#Fire up ten numbers
[ r.fire() for e in range(10) ]
```

# A function to truncate numbers
This lambda expression takes a float and truncates it to `num_digits`.

```python
num_digits = 2
truncateme = lambda x: int(x*10**num_digits)/(10**num_digits)
truncateme(0.2345)
```

# Histogram
Let's *bin* a large number of random numbers.

```python
bins = {x/10**num_digits:0 for x in range(100)}
rands = [r.fire() for e in range(100000)]
for _r in rands:
    bins[truncateme(_r)]+=1
values=bins.values()
values
```

## Verify pRNG
Let's calculate mean and variance. Also visually inspect the resulting histogram.

```python
mean = 0
var = 0
for val in values:
    mean += val
    var += val**2
mean = mean/len(values)
var = var/len(values) - mean**2
mean, var
```
