## Exercise solution

````python
import random
from math import sqrt

def f(x):
    return sqrt(1-x**2)

n = 0
N = 10000
for i in range(N):
    x = random.uniform(-1.,1.)
    y = random.uniform(0.,1.)
    if y < f(x):
        n += 1
res = n/N * 4
print(res)
````

alternative solution

````python
import random
from math import sqrt

n = 0
N = 10000
for i in range(N):
    x = random.uniform(0.,1.)
    y = random.uniform(0.,1.)
    if (x**2 + y**2) <1 :
        n += 1
res = n/N * 4
print(res)
````

### Important: Views or copy
A slice or reshape is a view, simply a re-organization of the same data in memory, thus changing one element changes the same element in all views

````python
a = np.arange(9)
````

````python
b = a.reshape((3,3))
````

````python
np.shares_memory(a,b)
````

````python
a[3] = -1
````

````python
b
````

````python
np.shares_memory(a,b)
````

````python
b = a.copy()
np.shares_memory(a,b)
````

### Boolean masks for extracting values
A typical operation done in your daily physics data analysis is to extract from an array the values that match a condition.
Consider an array of the energies of particles, and assume you want to use only the energies above a given threshold. Boolean masking comes at a rescue

````python
ene = np.random.exponential(size=10, scale=10.)
ene
````

````python
mask = ene > 3
mask
````

````python
ene[mask]
````

````python
ene[ene<3]
````

````python
ene[ene<3] = 0
ene
````

do you remember that in python function arguments are passed by copy?

````python
def f(x):
    x+=1

y = 1
f(y)
print(y)
````

this is NOT true for numpy arrays!

````python
def f(x):
    x+=1

y = np.array([1])
f(y)
print(y)
````

It's like numpy arrays are passed by reference, pay attention!

As in the function is not called the `copy()` method explicitly, the local variable is a view of the global.
